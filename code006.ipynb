```python
produtos = [ 
    {'nome': 'Pão','preço': 5.70, 'quantidade': 60},
    {'nome': 'Leite', 'preço': 5.00, 'quantidade': 200},
    {'nome': 'Água', 'preço':2.90, 'quantidade': 789},
    {'nome': 'Água tônica', 'preço':3.99, 'quantidade': 89}
]

```

```python

fat = 0

for produto in produtos:
    faturamento = produto['preço'] * produto['quantidade']
    fat_total += faturamento
    
    if faturamento >= 1000:
        print(f'{produto["nome"]} ALTO - seu Faturamento: R${faturamento:.2f}')
    elif faturamento <= 1000:
        print(f'{produto["nome"]} NORMAL - seu Faturamento: R${faturamento:.2f}')

print(f'faturamento total: R${fat_total:.2f}')
```
