```python
pedidos = [
    {"produto": "Notebook", "preco": 3500, "quantidade": 2},
    {"produto": "Mouse", "preco": 80, "quantidade": 5},
    {"produto": "Monitor", "preco": 1200, "quantidade": 3},
    {"produto": "Teclado", "preco": 150, "quantidade": 4}
]

faturamento_total = 0 
produtos_acima_1000 = 0
nome_dos_produtos_maior1000 = []
fatu_prod_mais_1000 = 0

for produto in pedidos:

    faturamento = produto["preco"] * produto["quantidade"] 
    faturamento_total += faturamento

    if faturamento > 1000:
        produtos_acima_1000 += 1 
        nome_dos_produtos_maior1000.append(produto['produto'])
        fatu_prod_mais_1000  +=  faturamento

        # Média do faturamento dos produtos acima de 1000

ticket_medio = fatu_prod_mais_1000 / produtos_acima_1000  



print(f'seu faturamento: {faturamento_total}')
print(produtos_acima_1000)
print(nome_dos_produtos_maior1000)
print(ticket_medio)
```

```python
# calculando o faturamento médio dos produtos

if faturamento > 0:
    media = fatu_prod_mais_1000 / produtos_acima_1000
print(media)
```
