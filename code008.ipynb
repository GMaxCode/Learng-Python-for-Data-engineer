```python
pedidos = [
    {"produto": "Notebook", "preco": 3500, "quantidade": 2},
    {"produto": "Mouse", "preco": 80, "quantidade": 5},
    {"produto": "Monitor", "preco": 1200, "quantidade": 3},
    {"produto": "Teclado", "preco": 150, "quantidade": 4}
]

# função de faturamento

def calcular_faturamento(preco,quantidade):
        return preco * quantidade


# função do Faturamento médio

def calcular_media(total,quantidade):
    return total / quantidade


# percorrendo o faturamento de cada produto da lista usando o for

fat_total = 0

for produto in pedidos:
        faturamento = calcular_faturamento(
                produto["preco"],
                produto["quantidade"])
        fat_total += faturamento
        print(produto["produto"],faturamento)

media = calcular_media(fat_total,len(pedidos))

print(f"Faturamento total da loja: {fat_total}")
print(f"Média: {media}")




```
