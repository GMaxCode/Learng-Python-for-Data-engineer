```python
pedidos = [
    {"produto": "Notebook", "preco": 3500},
    {"produto": "Mouse", "preco": 80},
    {"produto": "Monitor", "preco": 1200},
    {"produto": "Teclado", "preco": 150}
]


def calcular_total(valores):
    total = []

    for valor in valores:
        if valor['preco'] > 1000:
            total.append(valor) 

    return total

print(calcular_total(pedidos))
```
