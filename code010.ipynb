```python
import pandas as pd

produtos = [
    {"produto": "Notebook", "preco": 3500},
    {"produto": "Mouse", "preco": 80},
    {"produto": "Monitor", "preco": 1200}
]

df = pd.DataFrame(produtos)



produts_baratos = df[df["preco"] < 2000]['produto']



# print(df['preco'])

print(f'produtos baratos:\n{produts_baratos}')

# print(f'produtos com custo menor que 2.000:\n{prod_menor_2000}')
```

```python

vendas = [
    {"produto": "Notebook", "preco": 3500, "quantidade": 2},
    {"produto": "Mouse", "preco": 80, "quantidade": 5},
    {"produto": "Monitor", "preco": 1200, "quantidade": 3}
]

df = pd.DataFrame(vendas)

df['Faturamento'] = df['preco'] * df['quantidade']

df['preco_com_desconto'] = df['preco'] * 0.90

faturamento_total = df['Faturamento'].sum()

soma_precos = df['preco'].sum()


print(df)
print()
print(f'total preços: {soma_precos}')
print()
print(f'faturamento total: {faturamento_total}')
print()
print(df.info())
```

```python
# calculando a média

vendas = [
    {"produto": "Notebook", "preco": 3500, "quantidade": 2},
    {"produto": "Mouse", "preco": 80, "quantidade": 5},
    {"produto": "Monitor", "preco": 1200, "quantidade": 3}
]

df = pd.DataFrame(vendas)

media_precos = df['preco'].mean()
print(f'média: {media_precos:.2f}')
```

```python

# Identificando valores ausentes com .isna()

# True: o preço está ausente.
# False: o preço está preenchido.


import pandas as pd

vendas = [
    {"produto": "Notebook", "preco": 3500},
    {"produto": "Mouse", "preco": None},
    {"produto": "Monitor", "preco": 1200},
    {"produto": "Teclado", "preco": None}
]

df = pd.DataFrame(vendas)

precos_ausentes = df['preco'].isna().sum()

print(f'quantos ausentes: {precos_ausentes}')
```
